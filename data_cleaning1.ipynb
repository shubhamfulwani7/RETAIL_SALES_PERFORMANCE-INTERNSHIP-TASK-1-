```python
df = df.drop_duplicates()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

print(df.columns.tolist())
```

```python
display(df.head(10))
```

```python
display(df.describe(include="all").T)
```

```python
df["transaction_date"] = pd.to_datetime(
    df["transaction_date"],
    errors="coerce"
)
```

```python
print(
    "Invalid dates:",
    df["transaction_date"].isna().sum()
)
```

```python
numeric_columns = [
    "price_per_unit",
    "quantity",
    "total_spent"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

display(df[numeric_columns].describe())
```

```python
print("Quantity <= 0:",
      (df["quantity"] <= 0).sum())

print("Price <= 0:",
      (df["price_per_unit"] <= 0).sum())

print("Total Spent < 0:",
      (df["total_spent"] < 0).sum())
```

```python
df["calculated_total"] = (
    df["price_per_unit"] * df["quantity"]
)

df["difference"] = (
    df["total_spent"] - df["calculated_total"]
)

display(
    df[
        ["price_per_unit",
         "quantity",
         "total_spent",
         "calculated_total",
         "difference"]
    ].head(10)
)
```

```python
df["year"] = df["transaction_date"].dt.year

df["month_number"] = (
    df["transaction_date"].dt.month
)

df["month"] = (
    df["transaction_date"].dt.month_name()
)

df["quarter"] = (
    "Q" +
    df["transaction_date"]
    .dt.quarter
    .astype(str)
)

df["year_month"] = (
    df["transaction_date"]
    .dt.to_period("M")
    .astype(str)
)

df["day_of_week"] = (
    df["transaction_date"]
    .dt.day_name()
)
```

```python
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
display(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

display(df.head())
```

```python
df.to_csv(
    "retail_store_sales_cleaned.csv",
    index=False
)
```
